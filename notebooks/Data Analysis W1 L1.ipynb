## Data types

The most common Python data types are 
- Integers and floats

#### Numbers

Basic numerical operations are defined as usual. 

````python
x = 12
y = 1.2

print(f'x + y = {x + y}')
print(f'x - y = {x - y}')
print(f'x * y = {x * y:.3f}')
print(f'x/y = {x/y}')
print(f'x^y = {x**y:.5f}')
````

#### Strings
A Python string is an array of characters.
To declare and define a string, write the characters between single quotes.

````python
s1 = 'CS5100 Lab Week 1 '
s2 = 'Introduction to Python'

print(f's1 = {s1}')
print(f's2 = {s2}')
print(f's1 + s2 = {s1 + s2}')
````

#### Lists

Lists are collections of variables. Indexes start with 0. You can add items to a list with `append` and drop them with `pop`. You can create a list of lists.

````python
l_s = ['apple', 'banana', 'peach']
l_n = [1, 2, 3, 4, 5]

print(f'l_s = {l_s}')
print(f'l_n = {l_n}')

print(f'len(l_s) = {len(l_s)}')

print(f'l_n[3] = {l_n[3]}')
print(f'l_s[-1] = {l_s[-1]}')

print(f'l_n[1:4] = {l_n[1:4]}')

l_s.append('orange')
l_n.pop(2)
print(f'l_s = {l_s}')
print(f'l_n = {l_n}')

l_l = [l_s, l_n]
l_sn = l_s + l_n
print(f'l_l = {l_l}')
print(f'l_sn = {l_sn}')
````

#### Dictionaries

Dictionaries are collections of data indexed by arbitrary unique keys. You can add elements by specifying the new key and the new value.

````python
d1 = {'first_name': 'William', 'last_name': 'Louis', 'age': 5}
print(f'd1 = {d1}')
print(f"d1['first_name'] = {d1['first_name']}")

d1['address'] = 'Windsor Castle'
print(f'd1 = {d1}')
print(f"d1['address'] = {d1['address']}")

print(f'd1.keys() = {d1.keys()}')
print(f'd1.values() = {d1.values()}')

print(f'list(d1.keys()) = {list(d1.keys())}')
print(f'list(d1.values()) = {list(d1.values())}')
````

## NumPy

NumPy is an abbreviation for numerical Python. NumPy arrays are multi-dimensional arrays of data. 
To use them, you need to import the package `numpy`. Remember to always initialise the NumPy random generator for reproducibility.

````python
import numpy as np
np.random.seed(1234)
a = np.array([1, 2, 3, 4, 5])
b = np.array([[1, 2, 3, 4, 5], [2, 3, 4, 5, 6]])
c = np.arange(10)
z = np.zeros(2)
o = np.ones((2,3), dtype=int)
r = np.random.rand(3, 5)
x = a[3]
even = c[0:-1:2]
y = r[1, 2]
print(f'a = {a}')
print(f'b = {b}')
print(f'c = {c}')
print(f'z = {z}')
print(f'o = {o}')
print(f'r = {r}')
print(f'x = {x}')
print(f'even = {even}')
print(f'y = {y}')
````

Unlike lists, you can inspect the size of an array using the attribute `shape`,  compute the sum, mean and standard deviation of its elements, and perform elementwise or matrix operations between arrays of suitable sizes.

````python
a = np.random.rand(3)
b = np.array(range(1, 4))
print(f'a.shape = {a.shape}')
print(f'b.shape = {b.shape}')
print(f'np.sum(a) = {np.sum(a)}')
print(f'a.mean() = {a.mean()}')
print(f'np.std(b) = {np.std(b)}')
print(f'a = {a}')
print(f'b = {b}')
print(f'a + b = {a + b}')
print(f'a - b = {a - b}')
print(f'a * b = {a * b}')
print(f'a / b = {a / b}')
print(f'a^b = {a**b}')
print(f'a @ b = {a @ b}') # inner product
print(f'np.outer(a, b) = {np.outer(a, b)}') # outer product 
````

## Control Flow

Python control flow is similar to other programming languages. You create loops using `while` and `for` and conditional statements using `if` and `else`. 

````python
for i in range(10):
    print(f'{i}^{i} = {i**i}')
````

````python
# create a matrix
m = np.zeros([5, 5])
for i in range(5):
    j = 0
    while j < i:
        m[i, j] = i**j
        j = j + 1
print(f'm = {m}')
````

````python
# short cut
x = [i/2 for i in range(0, 40, 3)]
print('x = ', x)
````

## Matplotlib

Matplotlib is a library to visualise data. Again, you need to import it explicitly before using it. 

````python
from matplotlib import pyplot as plt
x = np.arange(-5, 5, .1)
y1 = x**2  - x + 1 
y2 = -x**2  + x - 1 
plt.plot(x, y1, '-g', alpha=0.8, label = 'convex')
plt.plot(x, y2, '-r', alpha=0.8, label = 'concave')
plt.xlabel('$x$')
plt.ylabel('$f(x)$')
plt.title('Two quadratic functions')
plt.legend()
plt.savefig('plot.pdf')
plt.show()
````

### Scatter plot
Scatter plots are useful to compare the values of two variables. For example, to see if a predictor, $X_j$, is correlated with the response, $Y$.

````python
df = pd.read_csv('carsDataAnalysis202627.csv')
x_name = 'weight'
y_name = 'acceleration'
z_name = 'mpg'
x = df[x_name].to_numpy()
y = df[y_name].to_numpy()
z = df[z_name].to_numpy()
response = [y, z]
response_name = [y_name, z_name]
fig, ax = plt.subplots(1, 2, figsize=[8, 4])
for i in [0, 1]:
    ax[i].plot(x, response[i], '.', alpha=.5, label='cars')
    ax[i].set_xlabel(x_name)
    ax[i].set_ylabel(response_name[i])
    a = x - x.mean()
    b = response[i] - response[i].mean()
    corr = (a @ b)/(np.sqrt(a @ a) * np.sqrt(b @ b)) 
    ax[i].set_title(f'{x_name}-{response_name[i]} correlation: {corr:.3f}')
    ax[i].legend()
fig.tight_layout()
plt.savefig('plot.pdf')
plt.show()
````
